```python
{
 "cells": [
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "# Time-Series Trend & Rolling Metrics Analysis\n",
    "This notebook demonstrates resampling, rolling averages, cumulative sums, and period-over-period changes to identify structural business trends hidden beneath daily noise."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "import pandas as pd\n",
    "import matplotlib.pyplot as plt\n",
    "\n",
    "# Load and setup index\n",
    "df = pd.read_csv('../data/raw/daily_revenue.csv')\n",
    "df['date'] = pd.to_datetime(df['date'])\n",
    "df_ts = df.set_index('date')\n",
    "\n",
    "print('Data Loaded Successfully.')"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 1. Resampling and Period-Over-Period Change"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "weekly_revenue = df_ts['revenue'].resample('W').sum()\n",
    "wow_change = weekly_revenue.pct_change() * 100\n",
    "\n",
    "print(\"Weekly Revenue:\\n\", weekly_revenue)\n",
    "print(\"\\nWeek-over-Week Change (%):\\n\", wow_change)"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 2. Rolling Windows (Smoothing the Noise)"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "df['revenue_ma7'] = df['revenue'].rolling(window=7).mean()\n",
    "\n",
    "plt.figure(figsize=(10, 5))\n",
    "plt.plot(df['date'], df['revenue'], label='Raw Daily', alpha=0.3, linestyle='--')\n",
    "plt.plot(df['date'], df['revenue_ma7'], label='7-Day MA', linewidth=2, color='blue')\n",
    "plt.title('Daily Revenue vs 7-Day Rolling Average')\n",
    "plt.legend()\n",
    "plt.show()"
   ]
  }
 ],
 "metadata": {
  "kernelspec": {
   "display_name": "Python 3",
   "language": "python",
   "name": "python3"
  }
 },
 "nbformat": 4,
 "nbformat_minor": 4
}
```
