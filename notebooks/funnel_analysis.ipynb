```python
{
 "cells": [
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "# Funnel Analysis & Drop-Off Detection\n",
    "This notebook analyzes the user journey to identify where the most significant drop-offs occur, allowing us to quantify lost revenue and prioritize engineering efforts."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "import pandas as pd\n",
    "import matplotlib.pyplot as plt\n",
    "\n",
    "# Define Funnel Stages\n",
    "stages = {\n",
    "    'Sign Up': 10000,\n",
    "    'Email Entered': 8000,\n",
    "    'Password Created': 6000,\n",
    "    'Email Verified': 5000,\n",
    "    'Payment Added': 4000,\n",
    "    'First Purchase': 2000\n",
    "}\n",
    "print(\"Funnel Data Loaded.\")"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 1. Calculate Drop-off Rates"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "stage_list = list(stages.values())\n",
    "stage_names = list(stages.keys())\n",
    "drop_off = []\n",
    "\n",
    "for i in range(len(stage_list) - 1):\n",
    "    users_before = stage_list[i]\n",
    "    users_after = stage_list[i+1]\n",
    "    users_lost = users_before - users_after\n",
    "    drop_off.append({\n",
    "        'from_stage': stage_names[i],\n",
    "        'to_stage': stage_names[i+1],\n",
    "        'users_lost': users_lost,\n",
    "        'drop_rate': f'{(users_lost / users_before) * 100:.1f}%'\n",
    "    })\n",
    "\n",
    "funnel_df = pd.DataFrame(drop_off)\n",
    "print(funnel_df)"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 2. Visualize the Funnel"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "fig, ax = plt.subplots(figsize=(10, 5))\n",
    "ax.bar(stages.keys(), stages.values(), color='#3b82f6')\n",
    "plt.xticks(rotation=45, ha='right')\n",
    "plt.title('User Journey Funnel')\n",
    "plt.show()"
   ]
  }
 ],
 "metadata": {
  "kernelspec": {
   "display_name": "Python 3",
   "language": "python",
   "name": "python3"
  }
 },
 "nbformat": 4,
 "nbformat_minor": 4
}
```
